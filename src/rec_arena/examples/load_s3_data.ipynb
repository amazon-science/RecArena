# Load Data from S3 Bucket

This notebook loads data from the example-bucket S3 bucket using AWS credentials.

```python
import pandas as pd
import os

# Set AWS profile for AWS authentication
os.environ["AWS_PROFILE"] = "default"
```

```python
# Load test data from S3
bucket_name = "example-bucket"
key = "RecArena/lastfm_1k/leave_one_out/train.parquet"
s3_path = f"s3://{bucket_name}/{key}"

print(f"Loading: {s3_path}")
df = pd.read_parquet(s3_path)
print(f"Successfully loaded! Shape: {df.shape}")
```

```python
average_lengths = df.groupby("user_id")["item_id"].count().mean()

print(average_lengths)
```
